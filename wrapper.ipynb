```python
class DCASEWrapper:
    """
    DCASE Challenge Wrapper that manages both development and evaluation datasets
    """
    
    def __init__(self, challenge_name, dictionary):
        """
        Initialize DCASE wrapper with challenge name
        
        Args:
            challenge_name (str): Name of the DCASE challenge (e.g., 'dcase2023_task6A')
            dictionary (dict): Dictionary containing challenge configurations
        """
        if challenge_name not in dictionary:
            raise ValueError(f"Challenge '{challenge_name}' not found in dictionary")
        
        self.challenge_name = challenge_name
        self.config = dictionary[challenge_name]
        
        # Initialize development dataset
        dev_config = self.config['development']
        self.development = soundata.initialize(
            dev_config['dataset'], 
            version=dev_config['version']
        )
        
        # Initialize evaluation dataset
        eval_config = self.config['evaluation']
        self.evaluation = soundata.initialize(
            eval_config['dataset'], 
            version=eval_config['version']
        )

    def download(self, **kwargs):
        """
        Download datasets with optimized remote selection
        
        Args:
            **kwargs: Additional arguments passed to the download method (e.g., force_overwrite, cleanup)
        """
        self.development.download(**kwargs)
        self.evaluation.download(**kwargs)
        
        
    def validate(self):
        """
        Validate datasets for specified splits
        """
        self.development.validate()
        self.evaluation.validate()
    
    def clip_ids(self, split='development'): # should I add for both instead of dev/eval?
        """
        Get clip IDs for specified split
        
        Args:
            split (str): 'development' or 'evaluation'
        """
        if split == 'development':
            return self.development.clip_ids
        elif split == 'evaluation':
            return self.evaluation.clip_ids
        else:
            raise ValueError("Split must be 'development' or 'evaluation'")
    
    def clip(self, clip_id, split='development'):
        """
        Get clip object for specified split
        
        Args:
            clip_id (str): ID of the clip
            split (str): 'development' or 'evaluation'
        """
        if split == 'development':
            return self.development.clip(clip_id)
        elif split == 'evaluation':
            return self.evaluation.clip(clip_id)
        else:
            raise ValueError("Split must be 'development' or 'evaluation'")
    
    def load_clips(self, split='development'):
        """
        Load all clips for specified split
        
        Args:
            split (str): 'development' or 'evaluation'
        """
        if split == 'development':
            return self.development.load_clips()
        elif split == 'evaluation':
            return self.evaluation.load_clips()
        else:
            raise ValueError("Split must be 'development' or 'evaluation'")
    
    def info(self):
        """
        Print information about the DCASE challenge and datasets
        """
        print(f"DCASE Challenge: {self.challenge_name}")
        print(f"Development dataset: {self.config['development']['dataset']} v{self.config['development']['version']}")
        print(f"Evaluation dataset: {self.config['evaluation']['dataset']} v{self.config['evaluation']['version']}")
        
        
        print("Development dataset info:")
        print(self.development)
        
        print("Evaluation dataset info:")
        print(self.evaluation)
    
    def __repr__(self):
        return f"DCASEWrapper(challenge='{self.challenge_name}', dev={self.config['development']['dataset']}, eval={self.config['evaluation']['dataset']})"


# Factory function to create DCASE wrapper
def datawrapper(challenge_name, dictionary=None):
    """
    Factory function to create DCASE wrapper
    
    Args:
        challenge_name (str): Name of the DCASE challenge
        dictionary (dict): Dictionary containing challenge configurations. 
    
    Returns:
        DCASEWrapper: Initialized wrapper instance
    """
    
    return DCASEWrapper(challenge_name, dictionary)
```
